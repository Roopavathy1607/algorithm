```python
import heapq

def prims_algorithm(adjacency_list):
    start_node = next(iter(adjacency_list))
    visited = {start_node}
    priority_queue = []
    mst_edges = set()
    mst_cost = 0
    for neighbor, weight in adjacency_list[start_node]:
        heapq.heappush(priority_queue, (weight, start_node, neighbor))

    while priority_queue:
        weight, u, v = heapq.heappop(priority_queue)

        if v not in visited:
            visited.add(v)
            mst_edges.add(tuple(sorted((u, v))))
            mst_cost += weight
            if v in adjacency_list:
                for neighbor, weight in adjacency_list[v]:
                    if neighbor not in visited:
                        heapq.heappush(priority_queue, (weight, v, neighbor))

    return mst_edges, mst_cost
if __name__ == "__main__":
    graph = {
        'A': [('B', 2), ('C', 6), ('D', 1)],
        'B': [('A', 2), ('C', 3)],
        'C': [('B', 3), ('A', 6), ('D', 5)],
        'D': [('A', 1), ('C', 5)]
    }

    mst_edges, mst_cost = prims_algorithm(graph)

    print("Minimum Cost Spanning Tree Edges:")
    for edge in mst_edges:
        print(edge)

    print(f"\nTotal Cost of MST: {mst_cost}")
```
