```python
import sys
def floyd_warshall(graph):
    vertices = set()
    for u, v in graph.keys():
        vertices.add(u)
        vertices.add(v)
    V = len(vertices)
    vertex_list = sorted(list(vertices))
    dist = {}
    for i in vertex_list:
        for j in vertex_list:
            if (i, j) in graph:
                dist[(i, j)] = graph[(i, j)]
            elif i == j:
                dist[(i, j)] = 0
            else:
                dist[(i, j)] = sys.maxsize
    for k in vertex_list:
        for i in vertex_list:
            for j in vertex_list:
                if dist[(i, k)] != sys.maxsize and dist[(k, j)] != sys.maxsize:
                    dist[(i, j)] = min(dist[(i, j)], dist[(i, k)] + dist[(k, j)])

    return dist
if __name__ == "__main__":
    graph_edges = {
        ('0', '1'): 3,
        ('0', '3'): 5,
        ('1', '2'): 2,
        ('3', '2'): 1,
        ('2', '4'): 4,
        ('4', '1'): -1
    }

    shortest_paths = floyd_warshall(graph_edges)

    print("Shortest paths between all pairs of vertices:")
    vertices = sorted(list(set(u for u, _ in graph_edges) | set(v for _, v in graph_edges)))
    for i in vertices:
        for j in vertices:
            distance = shortest_paths.get((i, j), sys.maxsize)
            if distance == sys.maxsize:
                print(f"Shortest path from {i} to {j}: Infinity")
            else:
                print(f"Shortest path from {i} to {j}: {distance}")
```

```python
import sys

def floyd_simple(graph):
    V = len(graph)
    dist = [row[:] for row in graph]

    for k in range(V):
        for i in range(V):
            for j in range(V):
                if dist[i][k] != sys.maxsize and dist[k][j] != sys.maxsize:
                    dist[i][j] = min(dist[i][j], dist[i][k] + dist[k][j])
    return dist
graph = [
    [0, 3, sys.maxsize, 5],
    [sys.maxsize, 0, 2, sys.maxsize],
    [sys.maxsize, sys.maxsize, 0, 1],
    [sys.maxsize, sys.maxsize, sys.maxsize, 0]
]

result = floyd_simple(graph)

print("Shortest paths between all pairs of vertices:")
for row in result:
    print([str(x) if x != sys.maxsize else "inf" for x in row])
```
