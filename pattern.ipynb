```python
def search(pat,txt):
  M=len(pat)
  N=len(txt)
  for i in range(N-M+1):
    j=0
    for j in range(M):
      if txt[i+j]!=pat[j]:
        break
    if j==M-1:
      print("pattern found at index",i)
txt="AABAACAADAABAAABAA"
pat="AABA"
search(pat,txt)
```
