```python
def depth_first_search(graph, start_node, visited=None, dfs_order=None):
    if visited is None:
        visited = set()
    if dfs_order is None:
        dfs_order = []

    visited.add(start_node)
    dfs_order.append(start_node)

    if start_node in graph:
        for neighbor in graph[start_node]:
            if neighbor not in visited:
                depth_first_search(graph, neighbor, visited, dfs_order)

    return dfs_order
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}

start_node = 'B'
dfs_result = depth_first_search(graph, start_node)
print(f"Depth-First Search starting from node '{start_node}': {dfs_result}")
```
