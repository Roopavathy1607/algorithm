```python
import itertools
distance = [
    [0, 10, 15, 20],
    [10, 0, 35, 25],
    [15, 35, 0, 30],
    [20, 25, 30, 0]
]
def tsp_brute_force(dist):
    n = len(dist)
    cities = list(range(n))
    min_path = None
    min_cost = float('inf')
    for perm in itertools.permutations(cities[1:]):
        path = [0] + list(perm) + [0]
        cost = sum(dist[path[i]][path[i+1]] for i in range(n))
        if cost < min_cost:
            min_cost = cost
            min_path = path
    return min_cost, min_path
def tsp_nearest_neighbor(dist):
    n = len(dist)
    visited = [False] * n
    path = [0]
    visited[0] = True
    total_cost = 0
    current = 0
    for _ in range(n - 1):
        next_city = None
        min_dist = float('inf')
        for j in range(n):
            if not visited[j] and dist[current][j] < min_dist:
                min_dist = dist[current][j]
                next_city = j
        path.append(next_city)
        visited[next_city] = True
        total_cost += min_dist
        current = next_city
    total_cost += dist[current][0]
    path.append(0)
    return total_cost, path
optimal_cost, optimal_path = tsp_brute_force(distance)
approx_cost, approx_path = tsp_nearest_neighbor(distance)
error = ((approx_cost - optimal_cost) / optimal_cost) * 100
print(f"Optimal Solution:\n  Path: {optimal_path}\n  Cost: {optimal_cost}")
print(f"Nearest Neighbor Approximation:\n  Path: {approx_path}\n  Cost: {approx_cost}")
print(f"Error in Approximation: {error:.2f}%")
```

```python
import itertools
distance = [
    [0, 10, 15, 20],
    [10, 0, 35, 25],
    [15, 35, 0, 30],
    [20, 25, 30, 0]
]
def tsp_optimal(distance):
    n = len(distance)
    min_cost = float('inf')
    best_path = []
    for perm in itertools.permutations(range(1, n)):
        path = [0] + list(perm) + [0]
        cost = sum(distance[path[i]][path[i+1]] for i in range(n))
        if cost < min_cost:
            min_cost = cost
            best_path = path
    return min_cost, best_path
def tsp_nearest_neighbor(distance):
    n = len(distance)
    visited = [False] * n
    path = [0]
    visited[0] = True
    cost = 0
    current = 0
    for _ in range(n - 1):
        next_city = None
        min_dist = float('inf')
        for j in range(n):
            if not visited[j] and distance[current][j] < min_dist:
                min_dist = distance[current][j]
                next_city = j
        visited[next_city] = True
        path.append(next_city)
        cost += min_dist
        current = next_city
    cost += distance[current][0]
    path.append(0)
    return cost, path
optimal_cost, optimal_path = tsp_optimal(distance)
approx_cost, approx_path = tsp_nearest_neighbor(distance)
error = ((approx_cost - optimal_cost) / optimal_cost) * 100
print("Optimal Path:", optimal_path)
print("Optimal Cost:", optimal_cost)
print("Approximate Path:", approx_path)
print("Approximate Cost:", approx_cost)
print(f"Error: {error:.2f}%")
```
